```python
# Data Structures in Pandas

# 1. Series — 1D Labeled Array
import pandas as pd

s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
# 2. DataFrame — 2D Labeled Table
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["Delhi", "Mumbai", "Bangalore"]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.index = ["a", "b", "c"]
df.columns = ["Name", "Age", "City"]
```
